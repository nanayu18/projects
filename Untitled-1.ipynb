```python
#load dataset onto csv file
import pandas as pd

df = pd.read_csv(
    "/Users/nathan/.cache/kagglehub/datasets/ranjitmandal/fraud-detection-dataset-csv/versions/1/Fraud Detection Dataset.csv"
)

df.sample(10)
```

```python
df.nunique()
```

```python
df.isna().sum()
```

```python
#replace missing values in numeric and category columns
numeric_cols = df.select_dtypes(include="number").columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())

category_cols = df.select_dtypes(exclude="number").columns
for col in category_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```
